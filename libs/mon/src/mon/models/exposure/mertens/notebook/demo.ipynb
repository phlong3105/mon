# Mertens Demo

Import

```python
from __future__ import annotations

import os

import cv2
import numpy as np
import torch

from mon import Path
from mon.models import Mertens
from mon.ops import write_image
```

Setup environment

```python
current_dir = Path(os.getcwd())
root_dir = current_dir.parents[0]
data_dir = current_dir / "demo"
output_dir = current_dir / "demo"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
results = {}

# Debug
print(f"Device: {device}")
```

Resolve I/O

```python
image_dir_name = "house"
# image_dir_name = "mask"

image_dir = data_dir / image_dir_name
image_files = sorted(list(image_dir.rglob("*")))

# Load data
images = [cv2.imread(str(f)) for f in image_files]
images = [cv2.cvtColor(img, cv2.COLOR_BGR2RGB) for img in images]
images = torch.Tensor(np.array(images))
images = images / 255.0  # Normalize to [0, 1]
images = images.permute(0, 3, 1, 2)  # Change to (N, C, H, W)
```

Process

```python
model = Mertens()
fusion = model(images)
```

Save

```python
save_file = output_dir / f"{image_dir.name}_debug" / f"{image_dir.name}_fusion.jpg"
write_image(fusion, save_file)
```
